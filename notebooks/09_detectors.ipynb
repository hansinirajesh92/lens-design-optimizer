```python
import numpy as np
import matplotlib.pyplot as plt
import sys
sys.path.append("..")
from lens.design import R2_for
from lens.detect import spherical, field_curvature, coma

best_form = [(0, 29.5, 1.5), (6, R2_for(29.5, 6, 50), 1.0)]
```

```python
print(spherical(best_form))
```

**spherical aberration detector.** compares where a straight-ahead ray through the edge of the f/5 opening (h = 5 mm) crosses the axis with where a ray near the center (h = 0.01 mm) crosses. for the best-form lens the edge ray crosses 0.52 mm earlier: spherical aberration of −0.52 mm. negative means the edge focuses closer to the lens.

```python
print(field_curvature(best_form))
```

**field curvature detector.** finds the best focus at 0° and at 10° (f/5) and subtracts them. for the best-form lens the 10° image is sharpest 2.55 mm closer to the lens than the center: field curvature of −2.55 mm. this matches the whole-field notebook (08), where the best focus moved from 52.3 mm to 49.75 mm.

```python
print(coma(best_form, 0))
print(coma(best_form, 10))
```

**coma detector.** traces three rays of an off-axis bundle (the two edges of the f/5 opening and the middle ray) to that angle's best focus. with no coma, the two edge rays land evenly either side of the middle ray. at 0° coma is exactly 0, as it must be by symmetry. at 10° the best-form lens's edge rays land, on average, 0.016 mm off-center from the middle ray: a small comet-shaped flare.

```python
pcl = [(0, 25, 1.5), (5, np.inf, 1.0)]
bfl = [(0, 29.5, 1.5), (6, R2_for(29.5, 6, 50), 1.0)]
winner = [(0, 100, 1.5), (6, R2_for(100, 6, 50), 1.0)]
print("spherical detectors:")
print("plano-convex lens")
print(spherical(pcl))
print("beft-form lens")
print(spherical(bfl))
print("winner lens")
print(spherical(winner))
print("firld curvature detectors:")
print("plano-convex lens")
print(field_curvature(pcl))
print("beft-form lens")
print(field_curvature(bfl))
print("winner lens")
print(field_curvature(winner, z_stop = -15))
print("coma detectors:")
print("plano-convex lens")
print(coma(pcl))
print("beft-form lens")
print(coma(bfl))
print("winner lens")
print(coma(winner, z_stop = -15))
```

all three detectors on three lenses (f/5, 10° for the off-axis ones). compare sizes, not signs: the sign only says which direction the blur goes.

| lens | spherical (mm) | field curvature (mm) | coma (mm) |
|---|---|---|---|
| plano-convex | −0.58 | −2.60 | −0.047 |
| best-form | −0.52 | −2.55 | +0.016 |
| winner (stop at −15 mm) | −1.39 | −0.85 | +0.008 |

each lens is best at what it was designed for. the best-form lens has the least spherical aberration, which is what it was optimized for (notebook 05). the winner has the flattest field (about 3x flatter), which is what it was optimized for (notebook 08), and as a bonus moving the stop also halved its coma. it pays for both with the most spherical aberration (about 2.7x the best-form lens). the plano-convex lens has the most coma, about 3x the best-form lens, tilted the other way.

no lens wins everything. the detectors show the trade-offs behind each design.

```python
for a in [2.5, 5, 10]:
    print(spherical(bfl, a))
```

spherical aberration of the best-form lens at three openings:

| opening | edge ray height | spherical (mm) |
|---|---|---|
| f/10 | 2.5 mm | −0.13 |
| f/5 | 5 mm | −0.52 |
| f/2.5 | 10 mm | −2.17 |

each time the opening doubles, spherical aberration gets about 4x bigger (×4.0, then ×4.1): it grows with the square of the opening. this detector measures the error along the axis. the blur on the sensor grows even faster, roughly with the cube (about 8x per doubling), because edge rays also travel more steeply. that's why closing the opening from f/2.5 to f/5 made the spot about 8x smaller in the whole-field notebook (08).

```python
for deg in [0, 5, 10]:
    print(deg, "degrees", "best-form lens:")
    print(coma(bfl, deg))
    print(deg, "degrees", "winner lens:")
    print(coma(winner, deg, z_stop = -15))
```

coma at three field angles (f/5):

| angle | best-form (mm) | winner, stop at −15 mm (mm) |
|---|---|---|
| 0° | 0 | 0 |
| 5° | 0.0063 | 0.0050 |
| 10° | 0.0156 | 0.0076 |

coma is exactly 0 at the center for both lenses, as it must be by symmetry, and grows toward the edge of the image. for the best-form lens it grows faster than the angle: doubling from 5° to 10° makes it about 2.5x bigger. the winner's coma grows much more slowly (about 1.5x), so the gap widens toward the edge: at 5° the two are close, at 10° the winner has half the coma. moving the stop in front of the lens helps most where coma is worst.

at its own best focus, the winner's spot (0.0214 mm) is 2.67x bigger than the best-form lens's (0.0080 mm), from the whole-field notebook (08). field curvature can't explain this: at each angle's own focus it adds no blur, and the winner has less of it anyway. the cause is spherical aberration, the blur that stays even at perfect focus. the winner's spherical aberration (−1.39 mm) is 2.66x the best-form lens's (−0.52 mm), almost exactly the same ratio as the spots. bending the lens to flatten the field brought back spherical aberration, and the detector measures exactly how much.
